# NYC Crime Dataset Cleaning
The purpose of this notebook is to clean the NYC crime dataset prior to EDA and project work. This includes removing unecessary characters/quotations around cell values, replacing empty rows with null, removing rows with missing data, and renaming columns.

#### Data Observations Part 1

* `(null)` appears as a literal string value for some columns (e.g. BORO_NM, PARKS_NM, CMPLNT_TO_TM, SUSP_AGE_GROUP, VIC_SEX, etc.) but these need to be changed to actual null values
* Some data/time columns are stored as a literal string value as well (e.g., CMPLNT_FR_DT, CMPLNT_TO_DT, RPT_DT, CMPLNT_FR_TM, CMPLNT_TO_TM) but should be stored as proper date, time, or datetime values
* Some of these date/time columns should be combined to get a proper date-time time stamp too (e.g., CMPLNT_FR_DT and CMPLNT_FR_TM)
* The coordinate columns (X_COORD_CD and Y_COORD_CD) contain values like "1,031,237" but these commas make them strings instead of numbers so the commas need to be stripped and the remaining string should be turned to numeric

#### Data Summary
Summary of columns and the data types, and different column values found for each

```python
def summarize_df(df, max_unique=15):
    rows = []
    for col in df.columns:
        series = df[col]
        total = len(series)
        
        null_count = series.isna().sum()
        # also catch string "(null)" values
        if series.dtype == object or pd.api.types.is_string_dtype(series):
            null_count += (series == '(null)').sum()
        pct_null = round(null_count / total * 100, 1)
        
        # data types found in column (handles mixed types)
        type_counts = series.dropna().map(type).value_counts()
        types_found = ", ".join([f"{t.__name__}({c})" for t, c in type_counts.items()])
        
        # unique values
        n_unique = series.nunique(dropna=True)
        if n_unique <= max_unique:
            sample_vals = series.dropna().unique().tolist()
            # remove string nulls
            sample_vals = [v for v in sample_vals if v != '(null)']
            values_summary = str(sample_vals)
        else:
            values_summary = f"{n_unique} distinct values"
        
        rows.append({
            "Column":        col,
            "Dtype":         str(series.dtype),
            "Types in Data": types_found,
            "Null/Empty":    f"{null_count} ({pct_null}%)",
            "Unique Count":  n_unique,
            "Values":        values_summary,
        })
    
    summary = pd.DataFrame(rows)
    return summary.style.set_properties(**{"text-align": "left"}) \
                        .set_table_styles([{"selector": "th", "props": [("text-align", "left")]}])

summarize_df(data_df)
```
